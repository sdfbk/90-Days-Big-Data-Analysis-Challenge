```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder\
        .appName('Basics')\
        .getOrCreate()
```

```python
netflix_df = spark.read.csv("/content/drive/MyDrive/netflix_titles.csv",header=True,inferSchema=True)
netflix_df.show()
```

```python
df_temp = netflix_df.createOrReplaceTempView("netflix_titles")
```

```python
sql_query = spark.sql("select * from netflix_titles")
sql_query.show()
```

```python
sql_q = spark.sql("select type as Movie_type,count(type) as Movie_Count from netflix_titles group by type")
sql_q.show()
```

```python
s = spark.sql("select * from netflix_titles where type = 'William Wyler'")
s.show()
```
