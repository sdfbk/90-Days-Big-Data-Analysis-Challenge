```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder\
        .appName("PySpark Basics")\
        .getOrCreate()
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
df = spark.read.csv("/content/drive/MyDrive/netflix_titles.csv",header=True,inferSchema=True)
```

```python
df.show(5)
```

```python
df.write.csv("/content/drive/MyDrive/netflix_copy.csv",mode = "overwrite")
```

```python
df_filter = df.filter((df["type"] == "Movie") & (df["release_year"].isin("2020","2021")))
df_filter.show()
```

```python
df_sel = df_filter.select("type","title","release_year", "country").groupBy("country").count()
df_sel.show()
```

```python
df_gr = df_sel.
df_gr
```
